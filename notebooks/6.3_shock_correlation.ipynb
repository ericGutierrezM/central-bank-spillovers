```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
banks = ["fed", "ecb", "boe"]
models = ["deepseekv3", "gemini25flash", "gpt-4o", "llama33", "mistrallarge_or", "qwen25_72b"]

for bank in banks:
    residuals = {}
    for m in models:
        path = f"../output/residuals/{bank}_{m}_residuals.csv" 
        df = pd.read_csv(path)
        residuals[m] = df.set_index("date")["shock"]

    shocks = pd.DataFrame(residuals)
    corr_matrix = shocks.corr(method='pearson')
    upper_tri = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
    mean_corr_fed = upper_tri.stack().mean()

    print(f"Mean pairwise correlation of {bank.upper()} communication shocks: {mean_corr_fed:.3f}")
```
